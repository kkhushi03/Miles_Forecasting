```python
df["flown miles lag_3"]  = df["flown miles"].shift(3)
df["flown miles lag_12"] = df["flown miles"].shift(12)
df["flown miles rolling mean 3"] = df["flown miles"].rolling(window=3).mean()
```

```python
#adding lags for booked miles
df["booked_miles_lag_1"] = df["Booked Miles"].shift(1)
df["booked_miles_lag_3"] = df["Booked Miles"].shift(3)
df["booked_miles_lag_6"] = df["Booked Miles"].shift(6)
```

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
TARGET = "flown miles"
FEATURES = [
    "Premier Redeemed",
    "accrued miles",
    'rpms',
    'Booked Miles',
    # 'gross passenger revenue',
    "flown miles lag_3",
    "flown miles lag_12",
    "booked_miles_lag_1",
    "booked_miles_lag_3",
    "booked_miles_lag_6",
    "flown miles rolling mean 3",
    "month_sin",
    "month_cos",
    # "accrued:redeemed",
    "booking_pressure",
    "load_factor_proxy",
    "Card Memb Redeemed"
]
```

```python
def walk_forward_split(df, train_size):
    train = df.iloc[:train_size]
    test = df.iloc[train_size:train_size + 1]
    return train, test
```

```python
df_model = df[FEATURES + [TARGET]].dropna().reset_index(drop=True)
```

```python
xgb_model = XGBRegressor(
    n_estimators=400,
    learning_rate=0.05,
    max_depth=2,
    subsample=0.7,
    colsample_bytree=0.7,
    objective="reg:squarederror",
    random_state=42
)
```

```python
predictions = []
actuals = []
initial_train_size = 36
for i in range(initial_train_size, len(df_model)):
    train, test = walk_forward_split(df_model, i)
    X_train = train[FEATURES]
    y_train = train[TARGET]
    X_test = test[FEATURES]
    y_test = test[TARGET]
    xgb_model.fit(X_train, y_train)
    y_pred = xgb_model.predict(X_test)[0]
    predictions.append(y_pred)
    actuals.append(y_test.values[0])
```

```python
mae = mean_absolute_error(actuals, predictions)
mse = mean_squared_error(actuals, predictions)
rmse = np.sqrt(mse)
print(f"MAE:  {mae:,.0f}")
print(f"RMSE: {rmse:,.0f}")
```

```python
#turning it into a percerntage error
actuals_arr = np.array(actuals)
preds_arr = np.array(predictions)
mape = np.mean(np.abs((actuals_arr - preds_arr) / actuals_arr)) * 100
print(f"MAPE: {mape:.2f}%")
```

```python
xgb_model.fit(df_model[FEATURES], df_model[TARGET])
```

```python
future_preds = []
df_future = df_model.copy()

for h in range(18):

    X_next = df_future[FEATURES].iloc[[-1]]
    next_pred = xgb_model.predict(X_next)[0]
    future_preds.append(next_pred)

    new_row = df_future.iloc[-1].copy()

    # update target
    new_row[TARGET] = next_pred

    # assume booked miles persist (or replace with booked forecast)
    new_row["Booked Miles"] = df_future["Booked Miles"].iloc[-1]

    # flown miles lags
    new_row["flown miles lag_3"] = df_future[TARGET].iloc[-3]
    new_row["flown miles lag_6"] = df_future[TARGET].iloc[-6]
    new_row["flown miles lag_12"] = df_future[TARGET].iloc[-12]

    # booked miles lags (correct reference column)
    new_row["booked_miles_lag_1"] = df_future["Booked Miles"].iloc[-1]
    new_row["booked_miles_lag_3"] = df_future["Booked Miles"].iloc[-3]
    new_row["booked_miles_lag_6"] = df_future["Booked Miles"].iloc[-6]

    # rolling mean
    new_row["flown miles rolling mean 3"] = (
        df_future[TARGET].iloc[-2:].sum() + next_pred
    ) / 3

    # time progression
    # new_row["time_step"] = new_row["time_step"] + 1

    df_future = pd.concat(
        [df_future, new_row.to_frame().T],
        ignore_index=True
    )
```

```python
future_preds = [int(x) for x in future_preds]
print(future_preds)
```
